# **Adding the Labels**

```python
import numpy as np

playerA_valid = (
    matches_df["playerA_svpt"].notna() &
    matches_df["playerA_1stWon"].notna() &
    matches_df["playerA_2ndWon"].notna() &
    (matches_df["playerA_svpt"] > 0)
)

matches_df["playerA_match_p"] = np.where(playerA_valid,
    (matches_df["playerA_1stWon"] + matches_df["playerA_2ndWon"]) / matches_df["playerA_svpt"],
    np.nan
)

playerB_valid = (
    matches_df["playerB_svpt"].notna() &
    matches_df["playerB_1stWon"].notna() &
    matches_df["playerB_2ndWon"].notna() &
    (matches_df["playerB_svpt"] > 0)
)

matches_df["playerB_match_p"] = np.where(playerB_valid,
    (matches_df["playerB_1stWon"] + matches_df["playerB_2ndWon"]) / matches_df["playerB_svpt"],
    np.nan
)
```

# **Reducing File Size**

```python
#check-1
X_train.info(memory_usage= 'deep')
```

```python
import pandas as pd
import numpy as np

def reduce_mem_usage(df):
    for col in df.columns:
        col_type = df[col].dtype

        if col_type == "bool" or "bool" in col_type.name:
            continue
        if col_type == "object" or col_type == "category" or "datetime" in col_type.name:
            continue

        c_min = df[col].min()
        c_max = df[col].max()

        if str(col_type)[:3] == 'int':
            if c_min >= np.iinfo(np.int8).min and c_max <= np.iinfo(np.int8).max:
                df[col] = df[col].astype(np.int8)
            elif c_min >= np.iinfo(np.int16).min and c_max <= np.iinfo(np.int16).max:
                df[col] = df[col].astype(np.int16)
            elif c_min >= np.iinfo(np.int32).min and c_max <= np.iinfo(np.int32).max:
                df[col] = df[col].astype(np.int32)
            else:
                df[col] = df[col].astype(np.int64)

        else:
            # Downcast binary-like floats (e.g. One-Hot encoded) to int8
            if c_min >= 0 and c_max <= 1 and df[col].isin([0.0, 1.0]).all():
                df[col] = df[col].astype(np.int8)
            elif c_min >= np.finfo(np.float32).min and c_max <= np.finfo(np.float32).max:
                df[col] = df[col].astype(np.float32)
            else:
                df[col] = df[col].astype(np.float64)

    return df

# matches_df = reduce_mem_usage(matches_df)
```
