# **Test/Validation/Train Split**

```python
matches_df_final_cleaned = matches_df.copy()
```

```python
# Split chronologically by complete tournament reference-date groups (70/10/20).
# No matches with the same tourney_date may cross a partition boundary.
match_dates = pd.to_datetime(matches_df["tourney_date"], errors="coerce")
if match_dates.isna().any():
    raise ValueError("Missing or invalid tournament dates: cannot make a temporal split.")

n = len(matches_df)
date_counts = match_dates.value_counts().sort_index()
if len(date_counts) < 3:
    raise ValueError("At least three distinct tournament dates are needed for train/val/test.")

cumulative = date_counts.cumsum()
train_candidates = cumulative.iloc[:-2]  # Leave dates for validation and test.
train_cutoff = (train_candidates - 0.70 * n).abs().idxmin()

val_candidates = cumulative.loc[cumulative.index > train_cutoff].iloc[:-1]
val_cutoff = (val_candidates - 0.80 * n).abs().idxmin()

train_mask = match_dates <= train_cutoff
val_mask = (match_dates > train_cutoff) & (match_dates <= val_cutoff)
test_mask = match_dates > val_cutoff

train_df = matches_df.loc[train_mask].copy()
val_df = matches_df.loc[val_mask].copy()
test_df = matches_df.loc[test_mask].copy()

# Store the split alongside source IDs and actual outcomes, never in predictors.
assert pd.to_datetime(match_metadata["tourney_date"]).reset_index(drop=True).equals(
    match_dates.reset_index(drop=True)
)
match_metadata["split"] = ""
match_metadata.loc[train_mask, "split"] = "train"
match_metadata.loc[val_mask, "split"] = "val"
match_metadata.loc[test_mask, "split"] = "test"
assert (match_metadata["split"] != "").all()
# An event ID must not cross split boundaries even if its recorded dates are inconsistent.
if "tourney_id" in match_metadata.columns:
    crossing = match_metadata.dropna(subset=["tourney_id"]).groupby("tourney_id")["split"].nunique()
    if (crossing > 1).any():
        raise AssertionError(f"Tournament IDs crossing temporal splits: {(crossing > 1).sum()}")
match_metadata.to_csv(DF_DIR / "matches_v1_1_metadata.csv", index=False)
```

```python
# Verify approximate 70/10/20 fractions and strictly ordered date groups.
for split_name, frame in [("Train", train_df), ("Validation", val_df), ("Test", test_df)]:
    print(f"{split_name}: {len(frame):,} ({len(frame) / n:.2%}), "
          f"dates {frame['tourney_date'].min()} to {frame['tourney_date'].max()}")

assert all(len(frame) > 0 for frame in (train_df, val_df, test_df))
assert len(train_df) + len(val_df) + len(test_df) == n
assert pd.to_datetime(train_df["tourney_date"]).max() < pd.to_datetime(val_df["tourney_date"]).min()
assert pd.to_datetime(val_df["tourney_date"]).max() < pd.to_datetime(test_df["tourney_date"]).min()
assert match_metadata["split"].value_counts().to_dict() == {
    "train": len(train_df), "val": len(val_df), "test": len(test_df)
}
```

### playerA_rank_points and playerB_rank_points

```python
#check-1
X_train[['playerA_rank_points', 'playerB_rank_points']].describe()
```

```python
# Training-fitted A/B-symmetric imputations for rankings and physical data.
IMPUTATION_VALUES = {"seed": float(MAX_SEED + 1)}

def pooled_training_stat(suffix, method="median"):
    pooled = pd.concat(
        [X_train[f"player{side}_{suffix}"] for side in ("A", "B")],
        ignore_index=True,
    ).replace([np.inf, -np.inf], np.nan).dropna()
    if pooled.empty:
        raise ValueError(f"No training observations available for {suffix}")
    if method == "max":
        return float(pooled.max())
    return float(pooled.median() if method == "median" else pooled.mean())

# Height is measured in centimetres. Implausible measurements are missing.
for frame in (X_train, X_val, X_test):
    for side in ("A", "B"):
        height_col = f"player{side}_ht"
        frame[height_col] = frame[height_col].where(frame[height_col].between(140, 230))
        rank_col = f"player{side}_rank_filled"
        frame[rank_col] = frame[rank_col].where(frame[rank_col].gt(0))

UNRANKED_RANK = pooled_training_stat("rank_filled", method="max") + 1
IMPUTATION_VALUES["rank_filled"] = UNRANKED_RANK
IMPUTATION_VALUES["rank_points"] = pooled_training_stat("rank_points")
IMPUTATION_VALUES["age"] = pooled_training_stat("age")
IMPUTATION_VALUES["ht"] = pooled_training_stat("ht")

for frame in (X_train, X_val, X_test):
    for side in ("A", "B"):
        for suffix in ("rank_filled", "rank_points", "age", "ht"):
            col = f"player{side}_{suffix}"
            frame[col] = frame[col].fillna(IMPUTATION_VALUES[suffix])
    # Recompute dependent features only after imputing the base columns.
    frame["rank_diff"] = frame["playerA_rank_filled"] - frame["playerB_rank_filled"]
    frame["rank_ratio"] = frame["playerA_rank_filled"] / frame["playerB_rank_filled"]
    frame["rank_points_diff"] = frame["playerA_rank_points"] - frame["playerB_rank_points"]
    frame["age_diff"] = frame["playerA_age"] - frame["playerB_age"]
    frame["height_diff"] = frame["playerA_ht"] - frame["playerB_ht"]
```

### 'p' Features

```python
#check-1
X_train[['playerA_matchup_p', 'playerB_matchup_p', 'playerA_surface_p', 'playerB_surface_p' ]].describe()
```

```python
# Surface and matchup serve probabilities: pooled training-only means.
for suffix in ("matchup_p", "surface_p"):
    fill_value = pooled_training_stat(suffix, method="mean")
    IMPUTATION_VALUES[f"serve_{suffix}"] = fill_value
    for frame in (X_train, X_val, X_test):
        for side in ("A", "B"):
            col = f"player{side}_{suffix}"
            frame[col] = frame[col].fillna(fill_value)
```

```python
# Missing rolling results reflect unresolved match ordering, not zero history.
# Preserve one indicator per ambiguous feature group and the allowed window size.
ROLLING_WINDOW_TARGETS = {"last10": 10, "last25": 25,
                          "surface10": 10, "surface25": 25}
for group, target in ROLLING_WINDOW_TARGETS.items():
    for side in ("A", "B"):
        prefix = f"player{side}_{group}"
        group_cols = [f"{prefix}_{suffix}" for suffix in ("wins", "losses", "matches", "pct")]
        for frame in (X_train, X_val, X_test):
            ambiguous = frame[group_cols].isna().any(axis=1)
            if not frame.loc[ambiguous, group_cols].isna().all(axis=1).all():
                raise ValueError(f"Unexpected partially missing rolling group: {prefix}")
            frame[f"{prefix}_missing"] = ambiguous.astype("int8")

    # Only the win-rate estimate is learned; missing groups receive the
    # nominal match count and a consistent wins/losses decomposition.
    fill_pct = pooled_training_stat(f"{group}_pct")
    IMPUTATION_VALUES[f"{group}_pct"] = fill_pct
    for frame in (X_train, X_val, X_test):
        for side in ("A", "B"):
            prefix = f"player{side}_{group}"
            mask = frame[f"{prefix}_missing"].eq(1)
            frame.loc[mask, f"{prefix}_matches"] = target
            frame.loc[mask, f"{prefix}_pct"] = fill_pct
            frame.loc[mask, f"{prefix}_wins"] = target * fill_pct
            frame.loc[mask, f"{prefix}_losses"] = target * (1 - fill_pct)

# A genuine last-five H2H tie has a known total but unknown win allocation.
# Use a neutral split and preserve an explicit ambiguity flag.
h2h_cols = ["playerA_last5_h2h_wins", "playerB_last5_h2h_wins", "last5_h2h_diff"]
for frame in (X_train, X_val, X_test):
    mask = frame[h2h_cols].isna().any(axis=1)
    frame["last5_h2h_missing"] = mask.astype("int8")
    frame.loc[mask, "last5_h2h_total"] = frame.loc[mask, "last5_h2h_total"].fillna(5)
    half = frame.loc[mask, "last5_h2h_total"] / 2
    frame.loc[mask, "playerA_last5_h2h_wins"] = half
    frame.loc[mask, "playerB_last5_h2h_wins"] = half
    frame.loc[mask, "last5_h2h_diff"] = 0.0
```

```python
import pandas as pd
import numpy as np

def reduce_mem_usage(df):
    for col in df.columns:
        col_type = df[col].dtype

        if col_type == "bool" or "bool" in col_type.name:
            continue
        if col_type == "object" or col_type == "category" or "datetime" in col_type.name:
            continue

        c_min = df[col].min()
        c_max = df[col].max()

        if str(col_type)[:3] == 'int':
            if c_min >= np.iinfo(np.int8).min and c_max <= np.iinfo(np.int8).max:
                df[col] = df[col].astype(np.int8)
            elif c_min >= np.iinfo(np.int16).min and c_max <= np.iinfo(np.int16).max:
                df[col] = df[col].astype(np.int16)
            elif c_min >= np.iinfo(np.int32).min and c_max <= np.iinfo(np.int32).max:
                df[col] = df[col].astype(np.int32)
            else:
                df[col] = df[col].astype(np.int64)

        else:
            # Downcast binary-like floats (e.g. One-Hot encoded) to int8
            if c_min >= 0 and c_max <= 1 and df[col].isin([0.0, 1.0]).all():
                df[col] = df[col].astype(np.int8)
            elif c_min >= np.finfo(np.float32).min and c_max <= np.finfo(np.float32).max:
                df[col] = df[col].astype(np.float32)
            else:
                df[col] = df[col].astype(np.float64)

    return df

# matches_df = reduce_mem_usage(matches_df)
```
