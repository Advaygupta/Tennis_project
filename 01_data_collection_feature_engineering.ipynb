# **Downloading The Data and Converting to Dataframe**

## *Downloading the Data*

```python
import requests
from pathlib import Path

save_folder = Path("data/tennis_atp")
save_folder.mkdir(exist_ok = True)

for year in range(1991,2027):
    file_types = ["atp_matches", "atp_matches_futures", "atp_matches_qual_chall"]

    for file_type in file_types:
        file_name = f"{file_type}_{year}.csv"

        url = f"https://raw.githubusercontent.com/JeffSackmann/tennis_atp/master/{file_name}"

        try:
            response = requests.get(url)

            if response.status_code == 200:
                with open(save_folder / file_name, "wb") as f:
                    f.write(response.content)
                    print(f"Downloaded {file_name}")

            else:
                print(f"Missing {file_name}")

        
        except Exception as e:
            print(e)
```

## *Combining and Converting to Dataframe*

```python
from pathlib import Path
import pandas as pd

DATA_DIR = Path("data/tennis_atp")

dfs = list()

for year in range(1991, 2027):
    file_types = ["atp_matches", "atp_matches_futures", "atp_matches_qual_chall"]

    for file in file_types:
        file_name = f"{file}_{year}.csv"
        file = DATA_DIR / file_name

        if file.exists():
            print(f"Loading {file_name}")
            df = pd.read_csv(file)
            dfs.append(df)
        else:
            print(f"Skipping missing file: {file_name}")

matches_df = pd.concat(dfs, axis = 0, ignore_index= True, sort= False)
```

```python
DF_DIR = Path("data/created_data")

matches_df.to_csv(DF_DIR / "matches_1991_2026_combined.csv", index = False)
```

```python
#creating features

from collections import defaultdict, deque
import numpy as np

#storage dictioneries
h2h_record = defaultdict(lambda: [0,0]) #key (playerA, playerB)
surface_h2h_record = defaultdict(lambda: [0,0]) #key ((playerA, playerB), surface)
last5_h2h_record = defaultdict(lambda: deque(maxlen= 5)) #key (playerA, playerB)
top20_record = defaultdict(lambda: [0,0]) #key player
hand_record = defaultdict(lambda: defaultdict(lambda: [0,0])) #key [player][opp_hand]

#feature_lists
h2h_playerA_wins = []
h2h_playerB_wins = []
h2h_total = []
h2h_diff = []

surface_h2h_playerA_wins = []
surface_h2h_playerB_wins = []
surface_h2h_total = []
surface_h2h_diff = []

last5_h2h_playerA_wins = []
last5_h2h_playerB_wins = []
last5_total = []
last5_h2h_diff = []

playerA_top20_wins = []
playerA_top20_losses = []
playerA_top20_win_pct = []
playerB_top20_wins = []
playerB_top20_losses = []
playerB_top20_win_pct = []

playerA_hand_wins = []
playerA_hand_losses = []
playerA_hand_win_pct = []
playerB_hand_wins = []
playerB_hand_losses = []
playerB_hand_win_pct = []


#loop to create the features
for row in matches_df.itertuples():
    playerA = row.playerA_id
    playerB = row.playerB_id
    winner = row.playerA_id if row.playerA_wins ==1 else row.playerB_id
    loser = row.playerB_id if row.playerA_wins == 1 else row.playerA_id
    surface = row.surface

    pair = (min(playerA, playerB), max(playerA, playerB))

    #---------------
    # OVERALL H2H
    #---------------
    h2h_record_pair_0 = h2h_record[pair][0]
    h2h_record_pair_1 = h2h_record[pair][1]

    if playerA == pair[0]:
        wins_A = h2h_record_pair_0
        wins_B = h2h_record_pair_1
    else:
        wins_A = h2h_record_pair_1
        wins_B = h2h_record_pair_0

    h2h_playerA_wins.append(wins_A)
    h2h_playerB_wins.append(wins_B)
    h2h_total.append(wins_A + wins_B)
    h2h_diff.append(wins_A - wins_B)


    #----------------
    # SURFACE H2H
    #----------------
    surface_pair = (pair, surface)

    surface_pair_0 = surface_h2h_record[surface_pair][0]
    surface_pair_1 = surface_h2h_record[surface_pair][1]

    if playerA == pair[0]:
        wins_A = surface_pair_0
        wins_B = surface_pair_1
    else:
        wins_A = surface_pair_1
        wins_B = surface_pair_0

    surface_h2h_playerA_wins.append(wins_A)
    surface_h2h_playerB_wins.append(wins_B)
    surface_h2h_diff.append(wins_A - wins_B)
    surface_h2h_total.append(wins_A + wins_B)


    #-----------------
    # LAST 5 H2H
    #-----------------
    history = last5_h2h_record[pair]

    pair0_wins = sum(history)
    pair1_wins = len(history) - pair0_wins

    if playerA == pair[0]:
        playerA_last5 = pair0_wins
        playerB_last5 = pair1_wins
    else:
        playerA_last5 = pair1_wins
        playerB_last5 = pair0_wins

    last5_h2h_playerA_wins.append(playerA_last5)
    last5_h2h_playerB_wins.append(playerB_last5)
    last5_total.append(len(history))
    last5_h2h_diff.append(playerA_last5 - playerB_last5)

    
    #----------------------------
    # PERFORMANCE AGAINST TOP20
    #----------------------------
    # for player A
    wins, losses = top20_record[playerA]

    playerA_top20_wins.append(wins)
    playerA_top20_losses.append(losses)
    playerA_top20_win_pct.append(wins/(wins+losses)) if wins+losses>0 else playerA_top20_win_pct.append(0.5)

    #for player B
    wins, losses = top20_record[playerB]

    playerB_top20_wins.append(wins)
    playerB_top20_losses.append(losses)
    playerB_top20_win_pct.append(wins/(wins+losses)) if wins+losses>0 else playerB_top20_win_pct.append(0.5)


    #--------------------------------
    # PERFORMANCE AGAINST OPP HAND
    #--------------------------------
    #for playerA
    opp_hand = row.playerB_hand

    wins, losses = hand_record[playerA][opp_hand]

    playerA_hand_wins.append(wins)
    playerA_hand_losses.append(losses)
    playerA_hand_win_pct.append(wins/(wins+losses)) if wins+losses > 0 else playerA_hand_win_pct.append(0.5)

    #for player B
    opp_hand = row.playerA_hand

    wins, losses = hand_record[playerB][opp_hand]

    playerB_hand_wins.append(wins)
    playerB_hand_losses.append(losses)
    playerB_hand_win_pct.append(wins/(wins+losses)) if wins+losses > 0 else playerB_hand_win_pct.append(0.5)





    #---------------
    # UPDATES
    #---------------
    #overall H2H
    if pair[0] == winner:
        h2h_record[pair][0] += 1
    else:
        h2h_record[pair][1] += 1

    #surface H2H
    if pair[0] == winner:
        surface_h2h_record[surface_pair][0] += 1
    else:
        surface_h2h_record[surface_pair][1] += 1

    #last 5 H2H
    if winner == pair[0]:
        history.append(1)
    else:
        history.append(0)

    #top 20
    if pd.notna(row.playerB_rank) and row.playerB_rank <=20:
        if winner==playerA:
            top20_record[playerA][0] += 1
        else:
            top20_record[playerA][1] += 1
    if pd.notna(row.playerA_rank) and row.playerA_rank <= 20:
        if winner == playerB:
            top20_record[playerB][0] += 1
        else:
            top20_record[playerB][1] +=1

    #hand
    if winner==playerA:
        hand_record[playerA][row.playerB_hand][0] +=1
        hand_record[playerB][row.playerA_hand][1] +=1
    else:
        hand_record[playerB][row.playerA_hand][0] += 1
        hand_record[playerA][row.playerB_hand][1] +=1



#attaching features to the dataframe
matches_df['playerA_h2h_wins'] = h2h_playerA_wins
matches_df['playerB_h2h_wins'] = h2h_playerB_wins
matches_df['h2h_total'] = h2h_total
matches_df['h2h_diff'] = h2h_diff

matches_df["playerA_surface_h2h_wins"] = surface_h2h_playerA_wins
matches_df["playerB_surface_h2h_wins"] = surface_h2h_playerB_wins
matches_df['surface_h2h_total'] = surface_h2h_total
matches_df['surface_h2h_diff'] = surface_h2h_diff

matches_df['playerA_last5_h2h_wins'] = last5_h2h_playerA_wins
matches_df['playerB_last5_h2h_wins'] = last5_h2h_playerB_wins
matches_df['last5_h2h_total'] = last5_total
matches_df['last5_h2h_diff'] = last5_h2h_diff

matches_df['playerA_top20_wins'] = playerA_top20_wins
matches_df['playerA_top20_losses'] = playerA_top20_losses
matches_df['playerA_top20_total'] = matches_df['playerA_top20_wins'] + matches_df['playerA_top20_losses'] #new
matches_df['playerA_top20_win_pct'] = playerA_top20_win_pct
matches_df['playerB_top20_wins'] = playerB_top20_wins
matches_df['playerB_top20_losses'] = playerB_top20_losses
matches_df['playerB_top20_total'] = matches_df['playerB_top20_wins'] + matches_df['playerB_top20_losses'] #new
matches_df['playerB_top20_win_pct'] = playerB_top20_win_pct

matches_df['playerA_hand_wins'] = playerA_hand_wins
matches_df['playerA_hand_losses'] = playerA_hand_losses
matches_df['playerA_hand_total'] = matches_df['playerA_hand_losses'] + matches_df['playerA_hand_wins'] #new
matches_df['playerA_hand_win_pct'] = playerA_hand_win_pct
matches_df['playerB_hand_wins'] = playerB_hand_wins
matches_df['playerB_hand_losses'] = playerB_hand_losses
matches_df['playerB_hand_total'] = matches_df['playerB_hand_losses'] + matches_df['playerB_hand_wins'] #new
matches_df['playerB_hand_win_pct'] = playerB_hand_win_pct
```

```python
#--------------
#Elo Parameters
#--------------
BASE_ELO = 1500.0
BASE_K = 32.0

# Surface blending weight
SURFACE_WEIGHT = 0.70
OVERALL_WEIGHT = 0.30

# Inactivity decay
DECAY_PER_DAY = 0.9995       # ~16% decay after one year


# ----------------------------------
# Tournament importance multipliers
# ----------------------------------
TOURNAMENT_MULTIPLIER = {
    "G": 2.0, #Grand Slam
    "M": 1.8, #ATP Masters 1000
    "A": 1.6, #ATP Tour 
    "C": 1.2, #Challenger

    # Everything else
    "F": 1.0,
    "S": 1.0,
    "D": 1.0,
    "O": 1.0,
    "15": 1.0,
    "25": 1.0,
}
```

```python
overall_elo = defaultdict(lambda: BASE_ELO)
surface_elo = defaultdict(lambda: None)
matches_played = defaultdict(int)
elo_history = defaultdict(list)
last_match_date = {}
last_match_date_by_surface = {}
```

```python
import re
def expected_score(rating_a, rating_b):
    """
    Computes the expected probability that player A defeats player B.
    """
    return 1.0 / (1.0 + 10 ** ((rating_b - rating_a) / 400))


def dynamic_k(matches_played):
    """
    Continuous K-factor that decreases as players become
    more established.
    """
    return BASE_K * (60 / (60 + matches_played))



def tournament_multiplier(level):
    """
    Returns the tournament importance multiplier.
    """
    return TOURNAMENT_MULTIPLIER.get(level, 1.0)


def best_of_five_multiplier(best_of):
    """
    Grand Slam men's matches (best-of-5) provide
    stronger evidence than best-of-3 matches.
    """
    if best_of == 5:
        return 1.15

    return 1.0



def margin_multiplier(score_string):
    """
    Computes a multiplier based on the total games won.

    Returns:
        1.0 if the score cannot be parsed.
    """

    if not isinstance(score_string, str):
        return 1.0

    score_string = score_string.upper()

    if "RET" in score_string or "W/O" in score_string:
        return 1.0

    winner_games = 0
    loser_games = 0

    sets = score_string.split()

    for s in sets:
        m = re.match(r"(\d+)-(\d+)", s)
        if m:
            winner_games += int(m.group(1))
            loser_games += int(m.group(2))

    margin = winner_games - loser_games

    if margin <= 0:
        return 1.0

    multiplier = np.log1p(margin)

    # Prevents extremely dominant wins from exploding Elo updates
    return min(multiplier, 2.5)


def apply_overall_inactivity_decay(player, current_date):
    """
    Applies rating decay towards BASE_ELO after
    30 days of inactivity overall.
    """

    if player not in last_match_date:
        return

    days_inactive = (current_date - last_match_date[player]).days

    if days_inactive <= 30:
        return

    decay = DECAY_PER_DAY ** (days_inactive - 30)

    # Overall Elo
    overall_elo[player] = BASE_ELO + (overall_elo[player] - BASE_ELO) * decay


def apply_surface_inactivity_decay(player, surface, current_date):
    """
    Applies rating decay towards BASE_ELO after
    30 days of inactivity on a specific surface.
    """
    key = (player, surface)
    if key not in last_match_date_by_surface:
        return

    days_inactive = (current_date - last_match_date_by_surface[key]).days

    if days_inactive <= 30:
        return

    decay = DECAY_PER_DAY ** (days_inactive - 30)

    # Surface Elo
    if surface_elo[key] is not None:
        surface_elo[key] = BASE_ELO + (surface_elo[key] - BASE_ELO) * decay


def get_surface_rating(player, surface):
    """
    Initializes a player's surface rating using
    their current overall Elo.
    """

    key = (player, surface)
    if surface_elo[key] is None:
        surface_elo[key] = overall_elo[player]

    return surface_elo[key]


def effective_k(matches_played, tourney_level, best_of, score):
    return (
        dynamic_k(matches_played)
        * tournament_multiplier(tourney_level)
        * best_of_five_multiplier(best_of)
        * margin_multiplier(score)
    )
```

```python
#--------------
#Feature Lists
#--------------

playerA_elo = []
playerB_elo = []
elo_diff = []

playerA_surface_elo = []
playerB_surface_elo = []
surface_elo_diff = []

playerA_blended_elo = []
playerB_blended_elo = []
blended_elo_diff = []

playerA_elo_ratio = []
playerB_elo_ratio = []

playerA_elo_momentum = []
playerB_elo_momentum = []

playerA_elo_volatility = []
playerB_elo_volatility = []


#----------
#Main Loop
#----------
for row in matches_df.itertuples():
    playerA = row.playerA_id
    playerB = row.playerB_id
    surface = row.surface
    current_date = row.tourney_date

    apply_overall_inactivity_decay(playerA, current_date)
    apply_overall_inactivity_decay(playerB, current_date)
    apply_surface_inactivity_decay(playerA, surface, current_date)
    apply_surface_inactivity_decay(playerB, surface, current_date)

    overall_A = overall_elo[playerA]
    overall_B = overall_elo[playerB]

    surface_A = get_surface_rating(playerA, surface)
    surface_B = get_surface_rating(playerB, surface)

    blended_A = SURFACE_WEIGHT*surface_A + OVERALL_WEIGHT*overall_A
    blended_B = SURFACE_WEIGHT*surface_B + OVERALL_WEIGHT*overall_B


    #storing features
    playerA_elo.append(overall_A)
    playerB_elo.append(overall_B)
    elo_diff.append(overall_A - overall_B)

    playerA_surface_elo.append(surface_A)
    playerB_surface_elo.append(surface_B)
    surface_elo_diff.append(surface_A - surface_B)

    playerA_blended_elo.append(blended_A)
    playerB_blended_elo.append(blended_B)
    blended_elo_diff.append(blended_A - blended_B)

    playerA_elo_ratio.append(overall_A / max(overall_B, 1))
    playerB_elo_ratio.append(overall_B / max(overall_A, 1))


    #elo momentum 
    if len(elo_history[playerA]) >= 10:
        playerA_momentum = overall_A - elo_history[playerA][-10]
    else:
        playerA_momentum = 0.0

    if len(elo_history[playerB]) >= 10:
        playerB_momentum = overall_B - elo_history[playerB][-10]
    else:
        playerB_momentum = 0.0
    playerA_elo_momentum.append(playerA_momentum)
    playerB_elo_momentum.append(playerB_momentum)



    #elo volatility
    if len(elo_history[playerA]) >= 20:
        playerA_volatility = np.std(elo_history[playerA][-20:])
    else:
        playerA_volatility = 0.0

    if len(elo_history[playerB]) >= 20:
        playerB_volatility = np.std(elo_history[playerB][-20:])
    else:
        playerB_volatility = 0.0

    playerA_elo_volatility.append(playerA_volatility)
    playerB_elo_volatility.append(playerB_volatility)



    #winner/loser
    if row.playerA_wins == 1:
        winner = playerA
        loser = playerB
        winner_rating = overall_A
        loser_rating = overall_B
        winner_surface = surface_A
        loser_surface = surface_B

    else:
        winner = playerB
        loser = playerA
        winner_rating = overall_B
        loser_rating = overall_A
        winner_surface = surface_B
        loser_surface = surface_A

    #expected scores
    winner_expected = expected_score(winner_rating, loser_rating)
    loser_expected = 1 - winner_expected

    winner_surface_expected = expected_score(winner_surface, loser_surface)
    loser_surface_expected = 1 - winner_surface_expected


    #effective k
    K_winner = (dynamic_k(matches_played[winner]) * tournament_multiplier(row.tourney_level) * best_of_five_multiplier(row.best_of) 
         * margin_multiplier(row.score))
    K_loser = (dynamic_k(matches_played[loser]) * tournament_multiplier(row.tourney_level) * best_of_five_multiplier(row.best_of) 
         * margin_multiplier(row.score))
    

    #updating elo
    overall_elo[winner] += K_winner * (1 - winner_expected)
    overall_elo[loser] += K_loser * (0 - loser_expected)

    surface_elo[(winner, surface)] += K_winner * (1 - winner_surface_expected)
    surface_elo[(loser, surface)] += K_loser * (0 - loser_surface_expected)

    elo_history[playerA].append(overall_elo[playerA])
    elo_history[playerB].append(overall_elo[playerB])


    matches_played[playerA] += 1
    matches_played[playerB] += 1

    last_match_date[playerA] = current_date
    last_match_date[playerB] = current_date
    last_match_date_by_surface[(playerA, surface)] = current_date
    last_match_date_by_surface[(playerB, surface)] = current_date
```

```python
from collections import defaultdict
import numpy as np
import pandas as pd

matchup_p_record = defaultdict(
    lambda: {
        "p1_won": 0,
        "p1_played": 0,
        "p2_won": 0,
        "p2_played": 0
    }
)

playerA_matchup_p = []
playerB_matchup_p = []

playerA_matchup_service_points = []
playerB_matchup_service_points = []


for row in matches_df.itertuples():
    playerA = row.playerA_id
    playerB = row.playerB_id

    pair = tuple(sorted((playerA, playerB)))

    record = matchup_p_record[pair]

    
    # PLAYER A / PLAYER B HISTORICAL MATCHUP P
    if pair[0] == playerA:
        A_won = record["p1_won"]
        A_played = record["p1_played"]

        B_won = record["p2_won"]
        B_played = record["p2_played"]

    else:
        A_won = record["p2_won"]
        A_played = record["p2_played"]

        B_won = record["p1_won"]
        B_played = record["p1_played"]


   
    # Player A features
    if A_played > 0:
        playerA_matchup_p.append(A_won / A_played)
    else:
        playerA_matchup_p.append(np.nan)

    playerA_matchup_service_points.append(A_played)


    
    # Player B features
    if B_played > 0:
        playerB_matchup_p.append(B_won / B_played)
    else:
        playerB_matchup_p.append(np.nan)

    playerB_matchup_service_points.append(B_played)


    # UPDATE HISTORY
    if (pd.notna(row.playerA_svpt) and pd.notna(row.playerB_svpt)):
        playerA_points_won = (row.playerA_1stWon + row.playerA_2ndWon)
        playerB_points_won = (row.playerB_1stWon + row.playerB_2ndWon)

        if pair[0] == playerA:
            record["p1_won"] += playerA_points_won
            record["p1_played"] += row.playerA_svpt

            record["p2_won"] += playerB_points_won
            record["p2_played"] += row.playerB_svpt

        else:
            record["p2_won"] += playerA_points_won
            record["p2_played"] += row.playerA_svpt

            record["p1_won"] += playerB_points_won
            record["p1_played"] += row.playerB_svpt



# ATTACH FEATURES
matches_df["playerA_matchup_p"] = playerA_matchup_p #label
matches_df["playerB_matchup_p"] = playerB_matchup_p #label

matches_df["playerA_matchup_service_points"] = (playerA_matchup_service_points)
matches_df["playerB_matchup_service_points"] = (playerB_matchup_service_points)
```

```python
from collections import defaultdict, deque

recent_history = defaultdict(deque)  # key = player_id


# feature lists
playerA_matches_7d = []
playerB_matches_7d = []

playerA_matches_30d = []
playerB_matches_30d = []

playerA_wins_7d = []
playerB_wins_7d = []

playerA_wins_30d = []
playerB_wins_30d = []

playerA_winpct_7d = []
playerB_winpct_7d = []

playerA_winpct_30d = []
playerB_winpct_30d = []


for row in matches_df.itertuples():
    playerA = row.playerA_id
    playerB = row.playerB_id

    current_date = row.tourney_date



    # PLAYER A FEATURES
    history = recent_history[playerA]
    
    # Optimize: Trim elements older than 30 days
    while history and (current_date - history[0][0]).days > 30:
        history.popleft()

    wins_7d = 0
    matches_7d = 0

    for match_date, result in history:
        if (current_date - match_date).days <= 7:
            matches_7d += 1
            wins_7d += result

    playerA_matches_7d.append(matches_7d)
    playerA_wins_7d.append(wins_7d)

    if matches_7d > 0:
        playerA_winpct_7d.append(wins_7d / matches_7d)
    else:
        playerA_winpct_7d.append(0.5)


    wins_30d = 0
    matches_30d = len(history)
    for _, result in history:
        wins_30d += result

    playerA_matches_30d.append(matches_30d)
    playerA_wins_30d.append(wins_30d)

    if matches_30d > 0:
        playerA_winpct_30d.append(wins_30d / matches_30d)
    else:
        playerA_winpct_30d.append(0.5)



    # PLAYER B FEATURES
    history = recent_history[playerB]
    
    # Optimize: Trim elements older than 30 days
    while history and (current_date - history[0][0]).days > 30:
        history.popleft()

    wins_7d = 0
    matches_7d = 0

    for match_date, result in history:
        if (current_date - match_date).days <= 7:

            matches_7d += 1
            wins_7d += result

    playerB_matches_7d.append(matches_7d)
    playerB_wins_7d.append(wins_7d)

    if matches_7d > 0:
        playerB_winpct_7d.append(wins_7d / matches_7d)
    else:
        playerB_winpct_7d.append(0.5)


    wins_30d = 0
    matches_30d = len(history)
    for _, result in history:
        wins_30d += result

    playerB_matches_30d.append(matches_30d)
    playerB_wins_30d.append(wins_30d)

    if matches_30d > 0:
        playerB_winpct_30d.append(wins_30d / matches_30d)
    else:
        playerB_winpct_30d.append(0.5)



    # UPDATE HISTORY
    if row.playerA_wins == 1:
        recent_history[playerA].append((current_date, 1))
        recent_history[playerB].append((current_date, 0))

    else:
        recent_history[playerA].append((current_date, 0))
        recent_history[playerB].append((current_date, 1))


matches_df["playerA_matches_7d"] = playerA_matches_7d
matches_df["playerB_matches_7d"] = playerB_matches_7d

matches_df["playerA_matches_30d"] = playerA_matches_30d
matches_df["playerB_matches_30d"] = playerB_matches_30d

matches_df["playerA_wins_7d"] = playerA_wins_7d
matches_df["playerB_wins_7d"] = playerB_wins_7d

matches_df["playerA_wins_30d"] = playerA_wins_30d
matches_df["playerB_wins_30d"] = playerB_wins_30d

matches_df["playerA_losses_7d"] = matches_df["playerA_matches_7d"] - matches_df["playerA_wins_7d"] #new
matches_df["playerB_losses_7d"] = matches_df["playerB_matches_7d"] - matches_df["playerB_wins_7d"] #new

matches_df["playerA_losses_30d"] = matches_df["playerA_matches_30d"] - matches_df["playerA_wins_30d"] #new
matches_df["playerB_losses_30d"] = matches_df["playerB_matches_30d"] - matches_df["playerB_wins_30d"] #new

matches_df["playerA_winpct_7d"] = playerA_winpct_7d
matches_df["playerB_winpct_7d"] = playerB_winpct_7d

matches_df["playerA_winpct_30d"] = playerA_winpct_30d
matches_df["playerB_winpct_30d"] = playerB_winpct_30d
```

```python
median_height = pd.concat([matches_df['playerA_ht'], matches_df['playerB_ht']]).median()

matches_df['playerA_ht'] = matches_df['playerA_ht'].fillna(median_height)
matches_df['playerB_ht'] = matches_df['playerB_ht'].fillna(median_height)

matches_df['height_diff'] = matches_df['playerA_ht'] - matches_df['playerB_ht']
```

## *Seed features*

```python
matches_df['playerA_seeded'] = matches_df['playerA_seed'].notna().astype('int8')
matches_df['playerB_seeded'] = matches_df['playerB_seed'].notna().astype('int8')
```

```python
matches_df['seed_diff'] = np.where(
    matches_df['playerA_seed'].notna() & matches_df['playerB_seed'].notna(), 
    matches_df['playerA_seed'] - matches_df['playerB_seed'], 
    np.nan
)
```

```python
matches_df['seed_diff_missing'] = matches_df['seed_diff'].isna().astype('int8')
```

## *Surface Specialization features*

```python
#win rate
matches_df['playerA_win_rate_specialization'] = matches_df["playerA_surface_career_pct"] - matches_df['playerA_career_pct']
matches_df['playerB_win_rate_specialization'] = matches_df['playerB_surface_career_pct'] - matches_df['playerB_career_pct']
```

```python
#elo
matches_df['playerA_elo_specialization'] = matches_df['playerA_surface_elo'] - matches_df['playerA_elo']
matches_df['playerB_elo_specialization'] = matches_df['playerB_surface_elo'] - matches_df['playerB_elo']
```

```python
#p
matches_df['playerA_serve_specialization'] = matches_df['playerA_surface_p'] - matches_df['playerA_career_p']
matches_df['playerB_serve_specialization'] = matches_df['playerB_surface_p'] - matches_df['playerB_career_p']
```

## *Experience*

```python
from collections import defaultdict

tournament_matches = defaultdict(int)                # key = (player_id, tourney_name)

level_matches = defaultdict(int)                     # key = (player_id, tourney_level)
level_wins = defaultdict(int)                        # key = (player_id, tourney_level)
level_losses = defaultdict(int)                      # key = (player_id, tourney_level)

tourney_minutes = defaultdict(float)                 # key = (player_id, tourney_id)
tourney_match_count = defaultdict(int)               # key = (player_id, tourney_id)

# Fill missing minutes with median match duration (median is robust to outliers)
median_minutes = matches_df['minutes'].median()
matches_df['minutes'] = matches_df['minutes'].fillna(median_minutes)


# FEATURE LISTS
playerA_tourney_matches = []
playerB_tourney_matches = []

playerA_level_matches = []
playerB_level_matches = []

playerA_level_wins = []
playerB_level_wins = []

playerA_level_losses = []
playerB_level_losses = []

playerA_minutes_tourney = []
playerB_minutes_tourney = []

playerA_avg_minutes_tourney = []
playerB_avg_minutes_tourney = []


for row in matches_df.itertuples():

    playerA = row.playerA_id
    playerB = row.playerB_id

    surface = row.surface
    tourney = row.tourney_name
    tourney_level = row.tourney_level
    tourney_id = row.tourney_id

    # -------------------------
    # TOTAL SAME TOURNAMENT MATCHES
    # -------------------------
    playerA_tourney_matches.append(tournament_matches[(playerA, tourney)])
    playerB_tourney_matches.append(tournament_matches[(playerB, tourney)])

    # -------------------------
    # TOTAL SAME TOURNEY LEVEL MATCHES
    # -------------------------
    playerA_level_matches.append(level_matches[(playerA, tourney_level)])
    playerB_level_matches.append(level_matches[(playerB, tourney_level)])

    # -------------------------
    # TOURNEY LEVEL WINS
    # -------------------------
    playerA_level_wins.append(level_wins[(playerA, tourney_level)])
    playerB_level_wins.append(level_wins[(playerB, tourney_level)])

    # -------------------------
    # TOURNEY LEVEL LOSSES
    # -------------------------
    playerA_level_losses.append(level_losses[(playerA, tourney_level)])
    playerB_level_losses.append(level_losses[(playerB, tourney_level)])

    # -------------------------
    # MINUTES PLAYED IN CURRENT TOURNAMENT
    # -------------------------
    playerA_minutes_tourney.append(tourney_minutes[(playerA, tourney_id)])
    playerB_minutes_tourney.append(tourney_minutes[(playerB, tourney_id)])

    # -------------------------
    # AVERAGE MINUTES IN CURRENT TOURNAMENT
    # -------------------------
    matches_played = tourney_match_count[(playerA, tourney_id)]
    minutes_played = tourney_minutes[(playerA, tourney_id)]

    if matches_played > 0:
        playerA_avg_minutes_tourney.append(minutes_played / matches_played)
    else:
        playerA_avg_minutes_tourney.append(0)

    matches_played = tourney_match_count[(playerB, tourney_id)]
    minutes_played = tourney_minutes[(playerB, tourney_id)]

    if matches_played > 0:
        playerB_avg_minutes_tourney.append(minutes_played / matches_played)
    else:
        playerB_avg_minutes_tourney.append(0)

    # -------------------------
    # UPDATE COUNTERS
    # -------------------------

    tournament_matches[(playerA, tourney)] += 1
    tournament_matches[(playerB, tourney)] += 1

    level_matches[(playerA, tourney_level)] += 1
    level_matches[(playerB, tourney_level)] += 1

    # Winner / loser updates
    if row.playerA_wins == 1:
        level_wins[(playerA, tourney_level)] += 1
        level_losses[(playerB, tourney_level)] += 1
    else:
        level_wins[(playerB, tourney_level)] += 1
        level_losses[(playerA, tourney_level)] += 1

    tourney_minutes[(playerA, tourney_id)] += row.minutes
    tourney_minutes[(playerB, tourney_id)] += row.minutes

    tourney_match_count[(playerA, tourney_id)] += 1
    tourney_match_count[(playerB, tourney_id)] += 1


# -------------------------
# ATTACH FEATURES
# -------------------------
matches_df["playerA_tourney_matches"] = playerA_tourney_matches
matches_df["playerB_tourney_matches"] = playerB_tourney_matches

matches_df["playerA_level_matches"] = playerA_level_matches
matches_df["playerB_level_matches"] = playerB_level_matches

matches_df["playerA_level_wins"] = playerA_level_wins #new
matches_df["playerB_level_wins"] = playerB_level_wins #new

matches_df["playerA_level_losses"] = playerA_level_losses #new
matches_df["playerB_level_losses"] = playerB_level_losses #new 

matches_df['playerA_level_win_pct'] = np.where(matches_df["playerA_level_matches"] > 0, matches_df["playerA_level_wins"]/matches_df["playerA_level_matches"], 0.5) #new
matches_df['playerB_level_win_pct'] = np.where(matches_df["playerB_level_matches"] > 0, matches_df["playerB_level_wins"]/matches_df["playerB_level_matches"], 0.5) #new

matches_df["playerA_minutes_tourney"] = playerA_minutes_tourney
matches_df["playerB_minutes_tourney"] = playerB_minutes_tourney

matches_df["playerA_avg_minutes_tourney"] = playerA_avg_minutes_tourney
matches_df["playerB_avg_minutes_tourney"] = playerB_avg_minutes_tourney
```
